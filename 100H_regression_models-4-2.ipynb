| Section | Regression target | How to read the result |
|---|---|---|
| **Model 1: time regression** | Best time in seconds | Difference in seconds |
| **Model 2: log-time regression** | Natural log of best time | Percentage difference in time |

Both models use the athlete’s outdoor eligibility year (first, second, third, or fourth year) and account for each athlete having a different starting speed. They also recognize when the same athlete appears in more than one season.
- Model 1: predicts their best season time in seconds.
- Model 2: predicts the log of their best season time.
Each model has its own chart and results. Hover over a point to see the athlete’s information. Each model is also tested to see how close its predictions are to actual times.

```python

import io, base64, zlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
plt.rcParams.update({'font.size':11,'axes.spines.top':False,'axes.spines.right':False})
EMBEDDED_DATA = ''
df = pd.read_csv(io.BytesIO(zlib.decompress(base64.b64decode(EMBEDDED_DATA))))
import numpy as np
import pandas as pd
from scipy.optimize import minimize_scalar
from scipy.stats import norm

def design(df):
    years=df['outdoor_eligibility_year'].to_numpy(int)
    return np.column_stack([np.ones(len(df)),*[years==j for j in (2,3,4)]]).astype(float)

def fit_mixed(df, log_time=False):
    """Gaussian random-intercept regression, fitted by profiled maximum likelihood.

    Covariance = residual_variance * I + athlete_variance * Z Z'.
    Uses the exact inverse/determinant of each athlete block (no approximation).
    """
    X=design(df);y=df.best_time_seconds.to_numpy(float)
    if log_time:y=np.log(y)
    ids,g=np.unique(df.athlete_id.astype(str),return_inverse=True)
    counts=np.bincount(g);n=len(y)
    def inverse_apply(A,lam):
        sums=np.zeros((len(ids),A.shape[1]));np.add.at(sums,g,A)
        return A-(lam/(1+lam*counts))[g,None]*sums[g]
    def evaluate(lam,details=False):
        ix=inverse_apply(X,lam);iy=inverse_apply(y[:,None],lam)[:,0]
        information=X.T@ix;beta=np.linalg.solve(information,X.T@iy)
        resid=y-X@beta;rss=float(resid@inverse_apply(resid[:,None],lam)[:,0])
        variance=rss/n
        objective=n*np.log(variance)+np.log1p(lam*counts).sum()
        if details:return beta,variance,np.linalg.inv(information),resid,objective
        return objective
    opt=minimize_scalar(lambda z:evaluate(np.exp(z)),bounds=(-18,18),method='bounded',options={'xatol':1e-10})
    lam=float(np.exp(opt.x))
    if evaluate(0)<=evaluate(lam):lam=0.0
    beta,var,info_inv,resid,obj=evaluate(lam,True)
    sums=np.bincount(g,weights=resid)
    effects=lam*sums/(1+lam*counts)
    posterior=var*lam/(1+lam*counts)
    fit={'beta':beta,'cov_beta':var*info_inv,'residual_variance':var,'athlete_variance':var*lam,
         'lambda':lam,'athlete_effects':dict(zip(ids,effects)),'posterior_variance':dict(zip(ids,posterior)),
         'log_time':log_time,'converged':bool(opt.success),'objective':obj,'n':n,'groups':len(ids)}
    fitted=X@beta+effects[g]
    fit['fitted_scale']=fitted;fit['residuals_scale']=y-fitted
    return fit

def predict_mixed(fit,df,mean=True):
    eta=design(df)@fit['beta']+np.array([fit['athlete_effects'].get(str(a),0.) for a in df.athlete_id])
    if not fit['log_time']:return eta
    if not mean:return np.exp(eta)
    # Expected seconds, integrating future residual and posterior athlete uncertainty.
    pv=np.array([fit['posterior_variance'].get(str(a),fit['athlete_variance']) for a in df.athlete_id])
    return np.exp(eta+0.5*(fit['residual_variance']+pv))

def coefficients(fit):
    b=fit['beta'];se=np.sqrt(np.diag(fit['cov_beta']));lo=b-1.96*se;hi=b+1.96*se
    frame=pd.DataFrame({'term':['First-year intercept','Year 2 vs 1','Year 3 vs 1','Year 4 vs 1'],
      'estimate':b,'standard_error':se,'lower_95':lo,'upper_95':hi,'p_value':2*norm.sf(abs(b/se))})
    if fit['log_time']:
        frame['percent_difference']=100*np.expm1(b);frame.loc[0,'percent_difference']=np.nan
        frame['percent_lower_95']=100*np.expm1(lo);frame['percent_upper_95']=100*np.expm1(hi)
        frame.loc[0,['percent_lower_95','percent_upper_95']]=np.nan
    return frame


LOG_MODEL = False
assert len(df)==705
assert not df.duplicated(['athlete_id','season']).any()
print('Loaded 705 athlete–season rows for 391 athletes.')
```

# Model 1: time regression

**Target:** `best_time_seconds`.

This model uses outdoor eligibility year—first, second, third, or fourth year—to predict an athlete’s best season time.

Athlete ID tells the model which results belong to the same person, so it can account for differences in athletes’ starting speeds.

The results are measured in seconds. A negative year difference means a faster time; a positive difference means a slower time.


```python
LOG_MODEL = False
model=fit_mixed(df, log_time=LOG_MODEL)
assert model['converged']
changes=100*np.expm1(model['beta'][1:]) if LOG_MODEL else model['beta'][1:]
unit='%' if LOG_MODEL else ' seconds'
results=pd.DataFrame({'Comparison':['Second year vs first','Third year vs first','Fourth year vs first'],
    'Result':[f'{abs(v):.2f}{unit} '+('faster' if v<0 else 'slower') for v in changes]})
display(results)

time_model = model
```

### The fitted regression equation
- Year2, Year3, and Year4 switch on the adjustment for that athlete’s eligibility year. Only the matching year equals 1; the others equal 0. For first year, all three equal 0.
- Athlete adjustment accounts for that athlete’s usual speed.
- Error is the difference between the actual time and the model’s prediction.

```python
# Show the actual fitted model, including its coefficients.
model = time_model
b = model['beta']
target = "best time in seconds"
print(f'{target} = {b[0]:.5f} {b[1]:+.5f}×Year2 {b[2]:+.5f}×Year3 {b[3]:+.5f}×Year4 + athlete adjustment + error')
display(pd.DataFrame({'Part':['First-year starting point','Year 2 vs 1','Year 3 vs 1','Year 4 vs 1'], 'Coefficient':b}))
```

### How well does the model predict?

The model learns from earlier seasons, then predicts times for a later season. Both models are tested on the **same 314 results from returning athletes**.

**Average error (MAE)** tells us how far the predictions are from the actual times. For example, an MAE of **0.44 seconds** means predictions are off by about 0.44 seconds on average. **Lower is better.**

We also compare the model with a simple guess: using the athlete’s previous season time as their next season time.

```python
LOG_MODEL = False
test_pieces=[]
for season in sorted(df.season.unique())[1:]:
    train=df[df.season<season]
    test=df[df.season==season].copy()
    if set(train.outdoor_eligibility_year)!={1,2,3,4}: continue
    train_model=fit_mixed(train, log_time=LOG_MODEL)
    assert train_model['converged']
    test['fitted_prediction']=predict_mixed(train_model,test)
    last=train.sort_values('season').groupby('athlete_id').best_time_seconds.last()
    test['previous_time']=test.athlete_id.map(last)
    test_pieces.append(test[test.previous_time.notna()])
test_results=pd.concat(test_pieces,ignore_index=True)
assert len(test_results)==314
mae=float(np.mean(abs(test_results.fitted_prediction-test_results.best_time_seconds)))
baseline_mae=float(np.mean(abs(test_results.previous_time-test_results.best_time_seconds)))
display(pd.DataFrame({'Method':['This model','Use previous season time'],
    'Average error (seconds)':[mae,baseline_mae]}))
print(f'Average prediction error: {mae:.3f} seconds across {len(test_results)} test results.')

model_1_mae = mae
```

# Model 2: log-time regression

This model predicts **log(time)** instead of time directly.

It uses the athlete’s **outdoor eligibility year**—first, second, third, or fourth year—and accounts for each athlete’s usual speed. **Athlete ID** connects the same person’s results across seasons.

The log scale lets us describe year differences as **percentages**. For example, a 2% decrease means a faster time.

Hover over a point to see the athlete’s time in **seconds**.


```python
LOG_MODEL = True
model=fit_mixed(df, log_time=LOG_MODEL)
assert model['converged']
changes=100*np.expm1(model['beta'][1:]) if LOG_MODEL else model['beta'][1:]
unit='%' if LOG_MODEL else ' seconds'
results=pd.DataFrame({'Comparison':['Second year vs first','Third year vs first','Fourth year vs first'],
    'Result':[f'{abs(v):.2f}{unit} '+('faster' if v<0 else 'slower') for v in changes]})
display(results)

log_model = model
```

### The fitted regression equation
- **Year2, Year3, and Year4** tell the model which eligibility year the athlete is in. The matching year gets a 1; the others get a 0. For first year, all three are 0.
- **First year** is the starting point for comparing years.
- **Athlete adjustment** accounts for each athlete’s usual speed.
- **Error** is the difference between the actual and predicted log(time).

```python
# Show the actual fitted model, including its coefficients.
model = log_model
b = model['beta']
target = "log(best time)"
print(f'{target} = {b[0]:.5f} {b[1]:+.5f}×Year2 {b[2]:+.5f}×Year3 {b[3]:+.5f}×Year4 + athlete adjustment + error')
display(pd.DataFrame({'Part':['First-year starting point','Year 2 vs 1','Year 3 vs 1','Year 4 vs 1'], 'Coefficient':b}))
```

### How well does this model predict?

The model learns from earlier seasons, then predicts times for a later season. Both models are tested on the **same 314 results from returning athletes**.

**Average error (MAE)** shows how far predictions are from actual times, measured in **seconds**. Lower is better.

We also compare the model with a simple guess: using each athlete’s previous season time as their next season time.

```python
LOG_MODEL = True
test_pieces=[]
for season in sorted(df.season.unique())[1:]:
    train=df[df.season<season]
    test=df[df.season==season].copy()
    if set(train.outdoor_eligibility_year)!={1,2,3,4}: continue
    train_model=fit_mixed(train, log_time=LOG_MODEL)
    assert train_model['converged']
    test['fitted_prediction']=predict_mixed(train_model,test)
    last=train.sort_values('season').groupby('athlete_id').best_time_seconds.last()
    test['previous_time']=test.athlete_id.map(last)
    test_pieces.append(test[test.previous_time.notna()])
test_results=pd.concat(test_pieces,ignore_index=True)
assert len(test_results)==314
mae=float(np.mean(abs(test_results.fitted_prediction-test_results.best_time_seconds)))
baseline_mae=float(np.mean(abs(test_results.previous_time-test_results.best_time_seconds)))
display(pd.DataFrame({'Method':['This model','Use previous season time'],
    'Average error (seconds)':[mae,baseline_mae]}))
print(f'Average prediction error: {mae:.3f} seconds across {len(test_results)} test results.')

model_2_mae = mae
```

The log-time model is slightly more accurate. Both models show faster best times in later eligibility years, but this does not prove that moving up a year causes athletes to improve.
